```python
'''Q2)Write a function note_breakup(amount)that returns a dictionary giving the minimum number of Rs.500,200,100,50,20,10 notes needed for an amount using // and % 
while looping through a list of denominations.Accept an amount, print only the denominations actually used in form "500 x 3", total number of notes and any unpaid
remainder not a multiple of 10''' 
def note_breakup(amount):
    denominations = [500, 200, 100, 50, 20, 10]
    result = {}
    remaining = amount
    total_notes = 0
    
    for d in denominations:
        count = remaining // d
        if count > 0:
            result[d] = count
            remaining %= d
            total_notes += count
            print(f"{d}x{count}")
            
    print(f"Total notes: {total_notes}")
    if remaining > 0:
        print(f"Unpaid remainder: {remaining}")
        
    return result

note_breakup(5678)
```
